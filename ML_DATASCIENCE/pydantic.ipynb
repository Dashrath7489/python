```python
from pydantic import BaseModel
class Student(BaseModel):
    Name : str
    Gender : str
    Age : int
    Location : str
st = Student(Name= "Chirage", Gender="Male",Age=32,Location="Indore")
print(st)
```

```python
class Employee(BaseModel):
    Name : str
    Age : int
    Salary : float

data = {
    "Name":"Baithel",
    "Age" : 22,
    "Salary": 100000
}

Emp = Employee(**data)
print(Emp)
```

```python
from pydantic import BaseModel, EmailStr

class Student(BaseModel):
    Name : str
    Gender : str ="Male"
    Age : int
    Location : str = "Indore"
    Email : EmailStr
    
st = Student(Name="Chirage",Gender="Male",Age=22,Location="Indore",Email="dashrath345@gmail.com")
print(st)
st.model_dump_json()
```
